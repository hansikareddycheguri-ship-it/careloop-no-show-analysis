```python
# Task 2.2 - checking the SQL clinic numbers with pandas groupby
import pandas as pd

appts = pd.read_csv("appointments_clean.csv")
clinics = pd.read_csv("clinics_enriched.csv")

merged = appts.merge(clinics, on="clinic_id")

by_clinic = merged.groupby(["clinic_name", "city"]).agg(
    appointments=("no_show", "count"),
    no_shows=("no_show", "sum"),
).reset_index()

by_clinic["no_show_pct"] = (100 * by_clinic["no_shows"] / by_clinic["appointments"]).round(1)
by_clinic = by_clinic.sort_values("no_show_pct", ascending=False)

print(by_clinic.to_string(index=False))
print("\nTotal appointments:", by_clinic["appointments"].sum())
print("Total no-shows:", by_clinic["no_shows"].sum())
```
